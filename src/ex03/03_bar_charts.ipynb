## Exercise 03 : Bar charts

```python
import pandas as pd
import sqlite3
```

### Make a connection to the database

```python
conn = sqlite3.connect('../data/checking-logs.sqlite')
```

### Run query

```python
query = """
SELECT
   uid, DATETIME(timestamp) AS commits
FROM checker
WHERE uid LIKE 'user%';
"""
df = pd.read_sql(sql=query, con=conn, parse_dates='commits')
df
```

### Create dataframe with date, weekday and hour columns

```python
df['date'] = df['commits'].dt.date
df['weekday'] = df['commits'].dt.dayofweek.apply(lambda x: 'weekend' if x in {5, 6} else 'working_day')
df['hour'] = pd.cut(x=df['commits'].dt.hour,
                       bins=list(range(25)),
                       labels=list(map(str, range(24))),
                       right=False, include_lowest=True)
df
```

### Group entries by date and hour

```python
df = df.iloc[:, 2:].groupby(['date', 'hour'], observed=True, as_index=False).value_counts()
df
```

### Create pivot table with average values of commits count

```python
df = df.pivot_table(index='hour', columns='weekday', values='count', aggfunc='mean', dropna=False, fill_value=0, observed=False)
df
```

### Create a bar charts

```python
bar_chart = df.plot(
    kind='bar',
    use_index=True,
    y = ['weekend',	'working_day'],
    figsize=(15, 8), 
    title='Commits per hour',
    rot=0,
    fontsize=8,   
    subplots=True,
    grid=False,
    legend=True, 
    color=['LimeGreen', 'CadetBlue']
)
```

```python
conn.close()
```
