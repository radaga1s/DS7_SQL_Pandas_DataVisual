## Exercise 05 : Boxplot

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

### Read the csv file

```python
df = pd.read_csv('../data/ab-test.csv', sep=',')
```

### Create dataframes for boxplots

```python
df_after = df[df['time'] == 'after']
df_before = df[df['time'] == 'before']
len(df_after)
```

### Create the boxplot

```python
after_graph, before_graph = plt.subplots(1, 2, figsize=(15,8), sharey=True)[1]

boxprops = dict(color='green', linewidth=3)
medianprops = dict(color='red', linewidth=2)
whiskerprops = dict(color='red', linewidth=1)
capprops = dict(color='red', linewidth=1)

df_after.boxplot(ax=after_graph, by='group', showmeans=True, boxprops=boxprops, medianprops=medianprops, whiskerprops=whiskerprops, capprops=capprops)
df_before.boxplot(ax=before_graph, by='group', showmeans=True, boxprops=boxprops, medianprops=medianprops, whiskerprops=whiskerprops, capprops=capprops)

after_graph.set_title('after')
before_graph.set_title('before')
plt.suptitle('Lag before labs deadlines. After and before the Newsfeed', fontsize=15)
plt.show()
```
